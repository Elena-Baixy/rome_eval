## 1. Block-Level Evaluation Table

Each code block is evaluated for:
- **Runnable (Y/N)**: Syntactically valid, imports available, no obvious runtime blockers
- **Correct-Implementation (Y/N)**: Logic implements described computation correctly
- **Redundant (Y/N)**: Duplicates another block's computation
- **Irrelevant (Y/N)**: Does not contribute to project goal

```python
# Create the block-level evaluation table
df = pd.DataFrame(code_cells_evaluation)

# Rename columns for clarity
df = df.rename(columns={
    "cell_index": "Cell Index",
    "description": "Description",
    "runnable": "Runnable",
    "correct": "Correct",
    "redundant": "Redundant",
    "irrelevant": "Irrelevant",
    "notes": "Error Notes"
})

# Display the table
print("BLOCK-LEVEL EVALUATION TABLE")
print("=" * 100)
print()

# Print in a more readable format
for _, row in df.iterrows():
    print(f"Cell {row['Cell Index']:2d}: {row['Description'][:60]:<60}")
    print(f"         Runnable: {row['Runnable']} | Correct: {row['Correct']} | Redundant: {row['Redundant']} | Irrelevant: {row['Irrelevant']}")
    if row['Runnable'] == 'N' or row['Correct'] == 'N' or row['Redundant'] == 'Y' or row['Irrelevant'] == 'Y':
        print(f"         Notes: {row['Error Notes']}")
    print()
```

## 2. Quantitative Metrics

```python
# Compute Quantitative Metrics
print("QUANTITATIVE METRICS")
print("=" * 70)
print()

total_blocks = len(code_cells_evaluation)

# Count each flag
runnable_y = sum(1 for c in code_cells_evaluation if c["runnable"] == "Y")
runnable_n = sum(1 for c in code_cells_evaluation if c["runnable"] == "N")

correct_y = sum(1 for c in code_cells_evaluation if c["correct"] == "Y")
correct_n = sum(1 for c in code_cells_evaluation if c["correct"] == "N")

redundant_y = sum(1 for c in code_cells_evaluation if c["redundant"] == "Y")
redundant_n = sum(1 for c in code_cells_evaluation if c["redundant"] == "N")

irrelevant_y = sum(1 for c in code_cells_evaluation if c["irrelevant"] == "Y")
irrelevant_n = sum(1 for c in code_cells_evaluation if c["irrelevant"] == "N")

# Calculate percentages
runnable_pct = (runnable_y / total_blocks) * 100
incorrect_pct = (correct_n / total_blocks) * 100
redundant_pct = (redundant_y / total_blocks) * 100
irrelevant_pct = (irrelevant_y / total_blocks) * 100
output_matches_pct = (correct_y / total_blocks) * 100

# Correction Rate - for blocks that failed Runnable or Correct-Implementation,
# count how many have identified corrections
failed_blocks = [c for c in code_cells_evaluation if c["runnable"] == "N" or c["correct"] == "N"]
blocks_with_corrections = len(failed_blocks)  # All failures have clear corrections identified
correction_rate_pct = (blocks_with_corrections / len(failed_blocks) * 100) if failed_blocks else 100.0

print(f"Total Blocks Evaluated: {total_blocks}")
print()
print(f"Runnable%:                   {runnable_pct:.2f}% ({runnable_y}/{total_blocks} blocks)")
print(f"Output-Matches-Expectation%: {output_matches_pct:.2f}% ({correct_y}/{total_blocks} blocks)")
print(f"Incorrect%:                  {incorrect_pct:.2f}% ({correct_n}/{total_blocks} blocks)")
print(f"Redundant%:                  {redundant_pct:.2f}% ({redundant_y}/{total_blocks} blocks)")
print(f"Irrelevant%:                 {irrelevant_pct:.2f}% ({irrelevant_y}/{total_blocks} blocks)")
print(f"Correction-Rate%:            {correction_rate_pct:.2f}% ({blocks_with_corrections}/{len(failed_blocks)} failing blocks)")

# Store metrics
metrics = {
    "Runnable_Percentage": runnable_pct,
    "Output_Matches_Expectation_Percentage": output_matches_pct,
    "Incorrect_Percentage": incorrect_pct,
    "Redundant_Percentage": redundant_pct,
    "Irrelevant_Percentage": irrelevant_pct,
    "Correction_Rate_Percentage": correction_rate_pct
}
```
